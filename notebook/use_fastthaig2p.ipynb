```python
from pythaitts import TTS
import IPython
```

```python
tts = TTS(pretrained="fastthaig2p")
```

```python
%%time
wave = tts.tts("ภาษาไทย ง่าย มาก มาก",return_type="waveform")
IPython.display.Audio(wave, rate=24000) # load a NumPy array
```
